```python
import torch
from ultralytics import YOLO
import os
```

```python
# --- CONFIGURATION ---
# Path to your preexisting weights file (e.g., 'runs/detect/train/weights/best.pt')
PREEXISTING_MODEL = '../yolov8_model/runs/detect/train/weights/best.pt' 

# Path to the data.yaml inside your filtered coral folder
DATA_YAML = 'coral_images/data.yaml'

# 1. Load your specific model
model = YOLO(PREEXISTING_MODEL)

# 2. Start Training (Fine-tuning)
results = model.train(
    data=DATA_YAML,
    epochs=50,         # 50 images is small, so 50-100 epochs is usually enough
    imgsz=640,         # Standard YOLO resolution
    batch=8,           # Small batch size for a small dataset
    name='coral_finetune',
    device=0 if torch.cuda.is_available() else 'cpu' # Uses GPU if available
)
```

```python
metrics = model.val()
metrics
```

```python
model.predict(
    source="coral_images/test/images",
    save=True,
    imgsz=640
)
```
