```python
import cv2
from ultralytics import YOLO
import torch

# --- CONFIGURATION ---
INPUT_VIDEO = 'Qualification 45 - 2025 Central Missouri Regional.mp4'  # Change this to your filename
MODEL_PATH = 'runs/detect/coral_finetune/weights/best.pt'
OUTPUT_NAME = 'cropped_inference_result_qual45.mp4'
# ---------------------

# 1. Load your best model
model = YOLO(MODEL_PATH)

# 2. Open the local video file
cap = cv2.VideoCapture(INPUT_VIDEO)

if not cap.isOpened():
    print(f"Error: Could not open video file {INPUT_VIDEO}")
    exit()

# Get original video properties
width  = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps    = cap.get(cv2.CAP_PROP_FPS)

# 3. Calculate the Crop (Bottom 2/5ths)
# If original is 1080p, crop_height will be 432 pixels
crop_height = int(height * (2/5))
start_y = height - crop_height

# 4. Setup Video Writer
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(OUTPUT_NAME, fourcc, fps, (width, crop_height))

print(f"Processing {INPUT_VIDEO}...")
frame_count = 0

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # Crop the frame: [y_start:y_end, x_start:x_end]
    cropped_frame = frame[start_y:height, 0:width]

    # Run YOLO inference
    # We use imgsz=640 to match your training configuration
    results = model.predict(
        source=cropped_frame, 
        conf=0.5, 
        device=0 if torch.cuda.is_available() else 'cpu', 
        verbose=False
    )

    # Draw boxes and labels on the cropped frame
    annotated_frame = results[0].plot()

    # Save the frame to the new video file
    out.write(annotated_frame)
    
    frame_count += 1
    if frame_count % 100 == 0:
        print(f"Processed {frame_count} frames...")

# Cleanup
cap.release()
out.release()
print(f"Finished! Result saved as: {OUTPUT_NAME}")
```
